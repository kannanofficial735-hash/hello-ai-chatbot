```python
import torch
import transformers

from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM,
)

MODEL_NAME = "google/flan-t5-small"
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Transformers version:", transformers.__version__)
print("Using device:", device)
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)
model = model.to(device)
model.eval()
print("FLAN-T5-small loaded successfully!")
```

```python
def ask(question):
    question = str(question).strip()
    if not question:
        return "Please type a question."
    prompt = ("Answer the following question " "in simple English: " + question)
    inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=512)
    inputs = inputs.to(device)
    with torch.inference_mode():
        output_ids = model.generate(**inputs, max_new_tokens=128, num_beams=2, do_sample=False)
    answer = tokenizer.decode(output_ids[0], skip_special_tokens=True)
    return answer.strip()

print(ask("What is artificial intelligence?"))
```

```python
while True:
    question = input("You: ")
    if question.strip().lower() in ["exit", "quit", "bye"]:
        print("Bot: Goodbye!")
        break
    answer = ask(question)
    print("Bot:", answer)
```
